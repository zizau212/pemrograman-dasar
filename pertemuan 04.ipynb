```python
# Latihan 1: Menentukan Tahun Kabisat

tahun = int(input("Masukkan tahun: "))

kabisat = (tahun % 400 == 0) or ((tahun % 4 == 0) and (tahun % 100 != 0))

print("Tahun Kabisat" if kabisat else "Bukan Tahun Kabisat")


# Latihan 2: Operasi Aritmatika

def operasi_aritmatika():
    # Input
    angka1 = float(input("Masukkan angka pertama: "))
    angka2 = float(input("Masukkan angka kedua: "))

    # Operasi aritmatika
    penjumlahan = angka1 + angka2
    pengurangan = angka1 - angka2
    perkalian = angka1 * angka2
    pembagian = angka1 / angka2

    # Output
    print("\nHasil Operasi:")
    print("Penjumlahan :", penjumlahan)
    print("Pengurangan :", pengurangan)
    print("Perkalian   :", perkalian)
    print("Pembagian   :", pembagian)


# Memanggil fungsi
operasi_aritmatika()























```

```python

harga_gorengan = int(input("Harga 1 gorengan: "))
jumlah_gorengan = int(input("Jumlah gorengan: "))

harga_esteh = int(input("Harga 1 es teh: "))
jumlah_esteh = int(input("Jumlah es teh: "))

uang = int(input("Uang yang dibayar: "))

# Operasi aritmatika
total_gorengan = harga_gorengan * jumlah_gorengan
total_esteh = harga_esteh * jumlah_esteh
total_belanja = total_gorengan + total_esteh
kembalian = uang - total_belanja

# Output

print("STRUK PEMBELIAN")
print("total gorengan :", total_gorengan)
print("Total es teh   :", total_esteh)
print("Total belanja  :", total_belanja)
print("Uang dibayar   :", uang)
print("Kembalian      :", kembalian)
```

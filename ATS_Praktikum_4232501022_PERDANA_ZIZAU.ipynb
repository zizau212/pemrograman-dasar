Soal 2. Energi Harian dan Capacity Factor PLTD (10 poin | CLO-1 | TP-1, TP-3)

```python
daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

# a. Menghitung energi harian
energi = daya_kw * jam

# Menampilkan energi harian
print(f"Energi harian = {energi} kWh")

# b. Menghitung capacity factor
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

# Menampilkan capacity factor dengan 2 angka desimal
print(f"Capacity factor = {capacity_factor:.2f}%")
```

Soal 8. Pengosongan Tangki Bahan Bakar (25 poin | CLO-1 | TP-5)

```python
# Inisialisasi awal volume tangki
volume = 100

# Perulangan while selama volume > 0
while volume > 0:
    volume -= 20  # Kurangi 20 liter
    print(f"Volume: {volume} liter")  # Cetak volume saat ini (setelah dikurangi)

    # Cek jika volume tepat mencapai 20 liter
    if volume == 20:
        print("WARNING: Hampir kosong!")

# Setelah loop selesai, cetak pesan tangki kosong
print("TANGKI KOSONG - Pengosongan berhenti.")
```

Soal 10. Pembacaan Suhu Bearing dengan Sensor Rusak (30 poin | CLO-1 | TP-5, TP-6)

```python
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

# Inisialisasi penghitung data valid
jumlah_valid = 0

# Menggunakan enumerate
for i, nilai in enumerate(data, start=1):
    try:
        # Ubah data menjadi float
        suhu = float(nilai)

        # Jika berhasil diubah, tambahkan jumlah data valid
        jumlah_valid += 1

        # Jika nilai melebihi 70.0, cetak peringatan
        if suhu > 70.0:
            print(f"Warning: Titik {i} suhu bearing tinggi!")

    except ValueError:
        # Jika gagal (ValueError), cetak pesan error dan hentikan perulangan dengan break
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break

# Tampilkan jumlah data valid setelah perulangan selesai
print(f"Jumlah data valid: {jumlah_valid}")
```

```python
# 1. Membuat list kosong untuk menyimpan data konsumsi listrik
data_listrik = []

# 2. Perulangan (for) untuk mengambil input selama 3 hari
for i in range(3):
    kwh = float(input("Masukkan konsumsi listrik harian (kWh): "))
    data_listrik.append(kwh) # Memasukkan angka ke dalam list

print("\n--- HASIL ANALISIS ENERGI ---")

# 3. Perulangan untuk membaca list dan Percabangan (if-elif-else) untuk cek kondisi
for kwh in data_listrik:
    if kwh > 50:
        print(f"Konsumsi {kwh} kWh -> Boros / Anomali")
    elif kwh >= 20:
        print(f"Konsumsi {kwh} kWh -> Normal")
    else:
        print(f"Konsumsi {kwh} kWh -> Hemat")
```

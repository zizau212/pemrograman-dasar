Aktivitas 6F

Prediksi, lalu jalankan

Lanjutkan tabel secara manual untuk dua elemen terakhir (75 dan 70). Berapa total akhir dan rata-ratanya? Berapa banyak skor yang 75 atau lebih tinggi? Kemudian jalankan blok tersebut.



Total akhir: 380

Rata-rata: 76.0 (diperoleh dari total 380 dibagi 5 elemen)

Banyak skor yang 75 atau lebih tinggi (>= 75): 3 skor (yaitu 80, 90, dan 75)

```python
scores = [80, 65, 90, 75, 70]

total = 0
high_scores = 0

for s in scores:
    total += s
    if s >= 75:
        high_scores += 1

average = total / len(scores)

print("Total akhir:", total)
print("Rata-rata:", average)
print("Skor >= 75:", high_scores)
```

Aktivitas 6G

Ubah jendela

Edit blok tersebut: jadikan rata-rata bergerak 5 hari (mulai perulangan pada indeks 4, iris temps[i - 4 : i + 1], bagi dengan 5). Berapa banyak rata-rata yang Anda dapatkan untuk 7 nilai? Mengapa?

```python
temps = [28, 31, 30, 33, 29, 27, 32]

# Daily change: start at index 1 because we look at i - 1
for i in range(1, len(temps)):
    change = temps[i] - temps[i - 1]
    print(f"day {i + 1}: {temps[i]} °C, change {change:+d}")

# 5-day moving average: window = the 5 values ending at index i
avg5 = []
for i in range(4, len(temps)):
    window = temps[i - 4 : i + 1]
    avg5.append(round(sum(window) / 5, 1))

print("5-day averages:", avg5)
print(len(temps), "values →", len(avg5), "averages")
```

Berapa banyak rata-rata yang Anda dapatkan untuk 7 nilai?
Jumlah rata-rata: 3 nilai rata-rata ([30.2, 30.0, 30.2]).Alasannya: Karena butuh 5 data berturut-turut (hari ke-1 sampai ke-5) untuk menghasilkan rata-rata pertama (berhenti di indeks ke-4). Dari total 7 data (len(temps) = 7), perulangan hanya berjalan pada indeks 4, 5, dan 6 (range(4, 7)), sehingga hanya ada 3 kelompok  yang bisa dihitung:
Indeks 0-4: [28, 31, 30, 33, 29]->30.2
Indeks 1-5: [31, 30, 33, 29, 27]->30.0
Indeks 2-6: [30, 33, 29, 27, 32]->30.2

```python
#Perdana Zizau
#4232501022
#21/09/2026
#Menampilkan hasil operasi aritmetika dua bilangan bulat
#soal no 1
a=int(input("memasukan bilangan bulat pertama"))
b=int(input("memasukan bilangan bulat kedua"))
print(f"{a} + {b}  = {a + b}")
print(f"{a} - {b}  = {a - b}")
print(f"{a} * {b}  = {a * b}")
if b != 0:
    print(f"{a} / {b}  = {a / b}")
    print(f"{a} // {b} = {a // b}")
    print(f"{a} % {b}  = {a % b}")
else:
    print("Pembagian (/, //, %) tidak bisa dilakukan karena bilangan kedua = 0")

#Perdana Zizau
#4232501022
#21/09/2026
#Menentukan apakah nilai ujian termasuk kategori lulus
#soal no 2
nilai = float(input("Masukkan nilai ujian (0-100): "))

print("Nilai ini termasuk kategori lulus (>= 60):", nilai >= 60)


#Perdana Zizau
#4232501022
#21/09/2026
#Menentukan apakah seseorang boleh mengemudi sendiri
#soal no 3
umur = int(input("Masukkan umur: "))
punya_sim = input("Punya SIM? (True/False): ").strip().lower() == "true"
didampingi = input("Didampingi instruktur? (True/False): ").strip().lower() == "true"

# Syarat utama: umur >= 17 dan punya SIM
boleh_mengemudi = umur >= 17 and punya_sim
print("Boleh mengemudi sendiri:", boleh_mengemudi)

# Latihan operator precedence:
# Aturan tambahan: umur harus >= 17, dan (punya SIM ATAU didampingi instruktur)
tanpa_kurung = umur >= 17 and punya_sim or didampingi     # dibaca: (umur >= 17 and punya_sim) or didampingi
dengan_kurung = umur >= 17 and (punya_sim or didampingi)  # sesuai maksud aturan

print("Tanpa kurung  :", tanpa_kurung)
print("Dengan kurung :", dengan_kurung)


#Perdana Zizau
#4232501022
#21/09/2026
#menghitung efisiensi termal sebuah unit
#soal no 4
energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

if energi_input > 0:
    efisiensi = (energi_output / energi_input) * 100
    print(f"Efisiensi termal unit pembangkit: {efisiensi:.2f}%")
else:
    print("Energi input harus lebih besar dari 0.")

#Perdana Zizau
#4232501022
#21/09/2026
#Menentukan apakah unit pembangkit perlu maintenance
#soal no 5

suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

perlu_maintenance = suhu_boiler > 550 or beban_operasi > 95

print(f"Suhu boiler > 550°C   : {suhu_boiler > 550}")
print(f"Beban operasi > 95%   : {beban_operasi > 95}")
print(f"Perlu maintenance     : {perlu_maintenance}")


```

3. Sebuah kasus bug nyata di forum Stack Overflow terjadi saat pemrogram menyaring data menggunakan pustaka Pandas dengan kode df[df['Usia'] > 18 & df['Nama'] == 'Budi'] [Medium]. Pemrogram berharap sistem menyaring data orang bernama Budi yang berusia di atas 18 tahun [Medium]. Namun, karena operator & memiliki prioritas lebih tinggi daripada operator relasional (> dan ==), Python justru mengevaluasi 18 & df['Nama'] terlebih dahulu [Medium]. Hal ini menyebabkan sistem eror (TypeError) atau menghasilkan data kosong karena urutan perhitungan yang salah [Medium].

Kode tersebut harus diperbaiki dengan menambahkan tanda kurung () di setiap pernyataan perbandingan. Tanda kurung akan memaksa Python untuk menyelesaikan pengecekan usia dan nama terlebih dahulu, baru kemudian menggabungkan keduanya cmiiw.

*kode yang salah (bug)
df[df['Usia'] > 18 & df['Nama'] == 'Budi']

*kode yang benar (fixed)
df[(df['Usia'] > 18) & (df['Nama'] == 'Budi')]

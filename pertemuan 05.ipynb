```python
# Soal 1 kalkulator faktorial 1-5
n = int(input("masukan bilangan 1-5"))

# TODO: Validasi input - jika n kurang dari 1 ATAU lebih dari 5, tampilkan pesan tidak valid
# Gunakan operator logika "or" untuk menggabungkan dua kondisi relasional
if n < 1 or n > 5:
    print("Input tidak valid. Masukkan bilangan antara 1 sampai 5.")

# TODO: Lengkapi setiap cabang elif dengan perhitungan faktorial menggunakan operator *
# Cabang untuk n == 1
elif n == 1:
    print("1! = 1")
elif n == 2:
    print("2! =", 1 * 2)
elif n == 3:
    print("3! =", 1 * 2 * 3)
elif n == 4:
    print("4! =", 1 * 2 * 3 * 4)
else:
    print("5! =", 1 * 2 * 3 * 4 * 5)

    # Program Klasifikasi Kualitas Daya

# Meminta input dari pengguna
faktor_daya = float(input("Masukkan faktor daya (0.0 - 1.0): "))
efisiensi = float(input("Masukkan efisiensi distribusi (%): "))

# Menghitung skor kualitas daya
skor = faktor_daya * efisiensi

# Menentukan klasifikasi kualitas daya
if skor >= 85 and faktor_daya >= 0.90:
    klasifikasi = "PRIMA - sistem dalam kondisi optimal"
elif skor >= 70 and faktor_daya >= 0.80:
    klasifikasi = "BAIK - sistem memadai, pantau rutin"
elif skor >= 50 or faktor_daya >= 0.75:
    klasifikasi = "CUKUP - ada ruang perbaikan"
else:
    klasifikasi = "BURUK - perlu perbaikan segera"

# Menampilkan hasil
print("\n--- HASIL KLASIFIKASI ---")
print(f"Faktor daya: {faktor_daya:.2f}")
print(f"Efisiensi distribusi: {efisiensi:.1f}%")
print(f"Skor kualitas daya: {skor:.2f}")
print(f"Klasifikasi: {klasifikasi}")


# Program Rekomendasi Mode Operasi PLTS

# 1. Input data
iradiasi = float(input("Masukkan iradiasi matahari (W/m²): "))
soc = float(input("Masukkan kapasitas baterai/SoC (%): "))

# 2. Menghitung estimasi daya panel
daya_panel = iradiasi * 0.18 * 10  # Efisiensi 18%, luas panel 10 m²

print(f"Estimasi daya panel: {daya_panel:.2f} Watt")

# 3. Menentukan mode operasi
if iradiasi > 600 and soc < 80:
    mode = "CHARGING"
    keterangan = "Iradiasi tinggi, baterai belum penuh, prioritaskan pengisian."

elif iradiasi > 600 and soc >= 80:
    mode = "GRID-EXPORT"
    keterangan = "Baterai sudah penuh, ekspor kelebihan daya ke jaringan."

elif iradiasi > 200 and soc < 20:
    mode = "NORMAL"
    keterangan = "Iradiasi dan daya baterai cukup, operasi mandiri."

elif iradiasi <= 200 and soc >= 50:
    mode = "BATTERY-ONLY"
    keterangan = "Iradiasi rendah, andalkan baterai."

else:
    mode = "EMERGENCY"
    keterangan = "Iradiasi rendah atau baterai hampir habis, kurangi beban."

# 4. Menampilkan hasil
print("Mode operasi:", mode)
print("Keterangan:", keterangan)
```

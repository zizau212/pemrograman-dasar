```python
#SOAL NO 1
jumlah_buku = 7
harga_buku = 18500
diskon_persen = 10
uang_budi = 150000

total = jumlah_buku * harga_buku
diskon = total * diskon_persen // 100
bayar = total - diskon
kembalian = uang_budi - bayar

per_orang = bayar // 4
sisa = bayar % 4
cukup = uang_budi >= bayar

print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", cukup)

#SOAL NO 2
jumlah_panel = 12
daya_panel = 450
jam = 5
efisiensi = 0.8
kebutuhan = 25
tarif = 1444.70

energi_harian = jumlah_panel * daya_panel * jam * efisiensi / 1000
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif
kekurangan = kebutuhan - energi_harian

mencukupi = energi_harian >= kebutuhan

print("Energi harian (kWh) :", round(energi_harian, 2))
print("Energi bulanan (kWh):", round(energi_bulanan, 2))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 1))
print("Kekurangan (kWh)    :", round(kekurangan, 2))
print("Mencukupi?          :", mencukupi)

#SOAL NO 3
daya_generator = 200
jam_operasi = 8
konsumsi_solar = 0.25
harga_solar = 13500
kapasitas_tangki = 1000

arus = 20
hambatan = 0.5

energi = daya_generator * jam_operasi
solar = energi * konsumsi_solar
biaya = solar * harga_solar

hari_penuh = kapasitas_tangki // solar
sisa_solar = kapasitas_tangki % solar

rugi_kabel = arus ** 2 * hambatan
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print("Energi (kWh)     :", energi)
print("Solar (liter)    :", float(solar))
print("Biaya (Rp)       :", float(biaya))
print("Tangki cukup     :", float(hari_penuh), "hari")
print("Sisa solar (L)   :", float(sisa_solar))
print("Rugi kabel (W)   :", float(rugi_kabel))
print("Rugi tinggi?     :", rugi_tinggi)
```
